```python
from langchain_core.tools import tool

@tool
def calculator(expression: str) -> str:
    """Calculate mathematical expressions such as 25*4 or 100/5."""
    import ast
    import operator

    operations = {
        ast.Add: operator.add,
        ast.Sub: operator.sub,
        ast.Mult: operator.mul,
        ast.Div: operator.truediv,
        ast.Pow: operator.pow,
        ast.USub: operator.neg,
    }

    def evaluate(node):
        if isinstance(node, ast.Constant) and isinstance(node.value, (int, float)):
            return node.value

        if isinstance(node, ast.BinOp) and type(node.op) in operations:
            return operations[type(node.op)](
                evaluate(node.left), evaluate(node.right)
            )

        if isinstance(node, ast.UnaryOp) and type(node.op) in operations:
            return operations[type(node.op)](evaluate(node.operand))

        raise ValueError("Unsupported mathematical expression")

    try:
        tree = ast.parse(expression, mode="eval")
        return str(evaluate(tree.body))
    except Exception as e:
        return f"Calculation error: {e}"

print("Calculator tool created successfully!")
```

```python
import sqlite3

# Create SQLite database
conn = sqlite3.connect("ai_agent_memory.db")
cursor = conn.cursor()

# Create memory table
cursor.execute("""
CREATE TABLE IF NOT EXISTS memories (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    information TEXT NOT NULL
)
""")

conn.commit()

print("SQLite database created successfully!")
print("Memory table is ready!")
```

```python
def save_memory(information):
    cursor.execute(
        "INSERT INTO memories (information) VALUES (?)",
        (information,)
    )
    conn.commit()
    return "Memory saved successfully!"

print(save_memory("My name is Alex"))
```

```python
from langchain_core.tools import tool

@tool
def memory_search(query: str) -> str:
    """Search previously stored user information in SQLite."""

    cursor.execute(
        "SELECT information FROM memories WHERE information LIKE ?",
        (f"%{query}%",)
    )

    results = cursor.fetchall()

    if results:
        return "\n".join(row[0] for row in results)

    return "No matching memories found."

print(memory_search.invoke({"query": "name"}))
```

```python
from langchain_core.messages import HumanMessage, ToolMessage

memory_tools = {
    "calculator": calculator,
    "memory_search": memory_search
}

def ask_agent(question):

    messages = [HumanMessage(content=question)]

    ai_response = llm_with_tools.invoke(messages)
    messages.append(ai_response)

    for tool_call in ai_response.tool_calls:

        selected_tool = memory_tools[tool_call["name"]]

        result = selected_tool.invoke(tool_call["args"])

        messages.append(
            ToolMessage(
                content=str(result),
                tool_call_id=tool_call["id"]
            )
        )

    if ai_response.tool_calls:
        final_response = llm_with_tools.invoke(messages)
        return final_response.content

    return ai_response.content

print("Memory-enabled AI agent created!")
```

```python
university_documents = [
    """
    University Attendance Policy:
    Students must attend lectures regularly.
    Attendance requirements are determined by the relevant course.
    Students should check their course guidelines for minimum attendance.
    """,

    """
    Assignment Policy:
    Students must submit assignments before the deadline.
    Late submissions may receive penalties according to course rules.
    Assignments should follow the lecturer's instructions.
    """,

    """
    Examination Policy:
    Students must bring their university identification card to examinations.
    Students should arrive before the examination starts.
    Academic misconduct is prohibited.
    """,

    """
    Library Policy:
    Students can borrow books using their university identification card.
    Borrowed books must be returned before the due date.
    Students should follow library rules.
    """
]

print("University knowledge base created!")
print("Total documents:", len(university_documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

print("Text splitter imported successfully!")
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50
)

chunks = []

for document in university_documents:
    chunks.extend(splitter.split_text(document))

print("Total text chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"\nChunk {i+1}:")
    print(chunk)
```

```python
from sentence_transformers import SentenceTransformer

# Load embedding model
embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

# Convert text chunks into embeddings
embeddings = embedding_model.encode(
    chunks,
    convert_to_numpy=True
)

print("Embeddings generated successfully!")
print("Embedding shape:", embeddings.shape)
```

```python
import faiss
import numpy as np

dimension = embeddings.shape[1]

vector_database = faiss.IndexFlatL2(dimension)

vector_database.add(
    np.array(embeddings, dtype="float32")
)

print("FAISS vector database created!")
print("Stored vectors:", vector_database.ntotal)
```

```python
from langchain_core.tools import tool

@tool
def rag_search(query: str) -> str:
    """Search the university knowledge base for relevant information."""

    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True
    ).astype("float32")

    distances, indices = vector_database.search(
        query_embedding, 3
    )

    results = []

    for index in indices[0]:
        if index != -1:
            results.append(chunks[index])

    return "\n\n".join(results) if results else "No relevant information found."

print("RAG Search Tool created successfully!")
```

```python
result = rag_search.invoke({
    "query": "What rules must students follow during examinations?"
})

print("Retrieved university information:")
print(result)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, ToolMessage

agent_tools = {
    "calculator": calculator,
    "memory_search": memory_search,
    "rag_search": rag_search
}

def university_agent(question):

    messages = [
        SystemMessage(content="""
        You are an intelligent university assistant.

        You have three tools:
        1. calculator: Use for mathematical calculations.
        2. memory_search: Use to retrieve previously saved user information.
        3. rag_search: Use to answer questions about university policies
           and information from the university knowledge base.

        Always use the appropriate tool when the question requires it.
        Base university policy answers on retrieved documents.
        If information is not available, say so honestly.
        """),
        HumanMessage(content=question)
    ]

    for step in range(5):

        ai_response = llm_with_tools.invoke(messages)
        messages.append(ai_response)

        if not ai_response.tool_calls:
            return ai_response.content

        for tool_call in ai_response.tool_calls:

            selected_tool = agent_tools.get(tool_call["name"])

            if selected_tool is None:
                result = "Tool not found."
            else:
                try:
                    result = selected_tool.invoke(tool_call["args"])
                except Exception as error:
                    result = f"Tool error: {error}"

            messages.append(
                ToolMessage(
                    content=str(result),
                    tool_call_id=tool_call["id"]
                )
            )

    return "Maximum tool execution steps reached."

print("Intelligent University Assistant created!")
```

```python
def university_agent(question):

    q = question.lower()

    # Memory questions
    if "my name" in q or "who am i" in q:
        result = memory_search.invoke({"query": "my name"})
        return result.splitlines()[0] if result else "No memory found."

    # University policy questions
    policy_keywords = [
        "examination", "exam", "attendance",
        "assignment", "university policy"
    ]

    if any(word in q for word in policy_keywords):
        result = rag_search.invoke({"query": question})
        return "According to the university knowledge base:\n" + result

    # Calculation questions
    messages = [
        SystemMessage(content="You are an intelligent university assistant. Use the calculator tool for calculations."),
        HumanMessage(content=question)
    ]

    for step in range(5):
        response = llm_with_tools.invoke(messages)

        if not response.tool_calls:
            return response.content

        messages.append(response)

        for call in response.tool_calls:
            result = agent_tools[call["name"]].invoke(call["args"])
            messages.append(
                ToolMessage(
                    content=str(result),
                    tool_call_id=call["id"]
                )
            )

    return "Maximum steps reached."
```

```python
from langchain_core.tools import tool
from datetime import datetime, time, timedelta
from zoneinfo import ZoneInfo

timezone_sl = ZoneInfo("Asia/Colombo")


@tool
def check_calendar_today(query: str) -> str:
    """Check today's events in the user's Google Calendar."""

    today = datetime.now(timezone_sl).date()

    start_of_day = datetime.combine(
        today, time.min, tzinfo=timezone_sl
    )

    end_of_day = start_of_day + timedelta(days=1)

    result = calendar_service.events().list(
        calendarId="primary",
        timeMin=start_of_day.isoformat(),
        timeMax=end_of_day.isoformat(),
        singleEvents=True,
        orderBy="startTime"
    ).execute()

    events = result.get("items", [])

    if not events:
        return "No events scheduled for today."

    output = []

    for event in events:
        start = event["start"].get(
            "dateTime",
            event["start"].get("date")
        )

        output.append(
            event.get("summary", "Untitled event") + " - " + start
        )

    return "\n".join(output)


print("Google Calendar tool created successfully!")
```

```python
from langchain_core.tools import tool
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

@tool
def create_calendar_event(title: str, date: str, start_time: str, duration_minutes: int) -> str:
    """
    Create a Google Calendar event after user approval.
    Date format: YYYY-MM-DD.
    Time format: HH:MM (24-hour).
    """

    tz = ZoneInfo("Asia/Colombo")

    start = datetime.strptime(
        f"{date} {start_time}", "%Y-%m-%d %H:%M"
    ).replace(tzinfo=tz)

    end = start + timedelta(minutes=duration_minutes)

    event = {
        "summary": title,
        "start": {
            "dateTime": start.isoformat(),
            "timeZone": "Asia/Colombo"
        },
        "end": {
            "dateTime": end.isoformat(),
            "timeZone": "Asia/Colombo"
        }
    }

    created = calendar_service.events().insert(
        calendarId="primary",
        body=event
    ).execute()

    return (
        "Event created successfully: "
        + created.get("summary", title)
        + "\n"
        + created.get("htmlLink", "")
    )

print("Calendar event creation tool created successfully!")
```

```python
from langchain_core.tools import tool

@tool
def search_student(student_id: str) -> str:
    """Search university student records using a student ID."""

    result = sheets_service.spreadsheets().values().get(
        spreadsheetId=SPREADSHEET_ID,
        range="Students!A1:D100"
    ).execute()

    rows = result.get("values", [])

    for row in rows[1:]:
        if row and row[0].lower() == student_id.lower():
            return (
                f"Student ID: {row[0]}\n"
                f"Name: {row[1]}\n"
                f"Course: {row[2]}\n"
                f"Email: {row[3]}"
            )

    return "Student not found."

print("Student search tool created successfully!")
```

```python
import re

# Save the previous assistant function
previous_university_agent = university_agent

def university_agent(question):
    q = question.lower()

    # Student record search
    student_match = re.search(
        r"\bst\d{3,}\b",
        q,
        re.IGNORECASE
    )

    if student_match and any(
        word in q for word in ["student", "find", "search", "details", "record"]
    ):
        return search_student.invoke({
            "student_id": student_match.group().upper()
        })

    # Use the existing assistant for other questions
    return previous_university_agent(question)

print("Student search integrated with university assistant!")
```

```python
import base64
from email.mime.text import MIMEText
from langchain_core.tools import tool

@tool
def send_university_email(to_email: str, subject: str, message: str) -> str:
    """Send a university notification email to a specified recipient."""

    email = MIMEText(message)
    email["to"] = to_email
    email["subject"] = subject

    encoded_message = base64.urlsafe_b64encode(
        email.as_bytes()
    ).decode()

    email_body = {"raw": encoded_message}

    sent_message = gmail_service.users().messages().send(
        userId="me",
        body=email_body
    ).execute()

    return (
        "Email sent successfully!\n"
        f"Recipient: {to_email}\n"
        f"Subject: {subject}\n"
        f"Message ID: {sent_message['id']}"
    )

print("Gmail email-sending tool created successfully!")
```

```python
# Store actions that are waiting for user approval
pending_action = None


def approve_pending_action(approval):

    global pending_action

    if pending_action is None:
        return "There are no pending actions."

    if approval.lower().strip() not in ["yes", "y", "approve"]:
        pending_action = None
        return "Action cancelled. Nothing was executed."

    action = pending_action
    pending_action = None

    # Create Google Calendar event
    if action["type"] == "calendar":
        return create_calendar_event.invoke({
            "title": action["title"],
            "date": action["date"],
            "start_time": action["start_time"],
            "duration_minutes": action["duration_minutes"]
        })

    # Send Gmail notification
    elif action["type"] == "email":
        return send_university_email.invoke({
            "to_email": action["to_email"],
            "subject": action["subject"],
            "message": action["message"]
        })

    return "Unknown action."


print("User approval system created successfully!")
```

```python
def request_calendar_approval(title, date, start_time, duration_minutes):
    global pending_action

    pending_action = {
        "type": "calendar",
        "title": title,
        "date": date,
        "start_time": start_time,
        "duration_minutes": duration_minutes
    }

    return (
        "CALENDAR EVENT REQUIRES APPROVAL\n"
        f"Title: {title}\n"
        f"Date: {date}\n"
        f"Start time: {start_time}\n"
        f"Duration: {duration_minutes} minutes\n"
        "Approve this action? (yes/no)"
    )


def request_email_approval(to_email, subject, message):
    global pending_action

    pending_action = {
        "type": "email",
        "to_email": to_email,
        "subject": subject,
        "message": message
    }

    return (
        "EMAIL REQUIRES APPROVAL\n"
        f"Recipient: {to_email}\n"
        f"Subject: {subject}\n"
        f"Message: {message}\n"
        "Approve this action? (yes/no)"
    )


print("Calendar and email approval requests created successfully!")
```

```python
def handle_approval_response(user_response):
    response = user_response.lower().strip()

    if response in ["yes", "y", "approve"]:
        return approve_pending_action("yes")

    if response in ["no", "n", "cancel"]:
        return approve_pending_action("no")

    return "Please answer yes to approve or no to cancel."


print("Approval response handler created successfully!")
```

```python
def final_university_assistant(question):

    global pending_action

    q = question.lower().strip()

    # Handle pending approval
    if pending_action is not None:
        if q in ["yes", "y", "approve", "no", "n", "cancel"]:
            return handle_approval_response(q)

    # Handle normal university questions
    return university_agent(question)


print("Final university assistant interface created successfully!")
```

```python
def final_university_assistant(question):

    global pending_action

    q = question.strip()

    # Handle approval responses
    if pending_action is not None:
        if q.lower() in ["yes", "y", "approve", "no", "n", "cancel"]:
            return handle_approval_response(q)

    # Calendar request format:
    # CREATE EVENT | Title | YYYY-MM-DD | HH:MM | Duration

    if q.upper().startswith("CREATE EVENT |"):
        parts = q.split("|")

        if len(parts) != 5:
            return "Use: CREATE EVENT | Title | YYYY-MM-DD | HH:MM | Duration"

        return request_calendar_approval(
            parts[1].strip(),
            parts[2].strip(),
            parts[3].strip(),
            int(parts[4].strip())
        )

    # Email request format:
    # SEND EMAIL | Recipient | Subject | Message

    if q.upper().startswith("SEND EMAIL |"):
        parts = q.split("|", 3)

        if len(parts) != 4:
            return "Use: SEND EMAIL | Recipient | Subject | Message"

        return request_email_approval(
            parts[1].strip(),
            parts[2].strip(),
            parts[3].strip()
        )

    # Handle all other questions
    return university_agent(question)


print("Calendar and Gmail connected to the main assistant!")
```
